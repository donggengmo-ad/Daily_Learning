```python
import pandas as pd
df = pd.DataFrame({
    'subject': ['c', '军理', '军训', '体育', '高数', '近代史', '线代', '英语', '国安'],
    'weight': [3.5, 2, 2, 1, 5.5, 3, 3, 4, 1],
    'score': [100, 98, 85, 88, 84, 88, 87, 92, 97],
    'contribute': [0, 0, 0, 0, 0, 0, 0, 0, 0],
    'sum': [0, 0, 0, 0, 0, 0, 0, 0, 0]
})

df['contribute'] = df['score'] * df['weight'] / df['weight'].sum()
df['sum'] = df['contribute'].cumsum()

df
```
