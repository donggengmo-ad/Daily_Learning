```python
import pandas as pd
import numpy as np

def deg_to_float(x:float):
    x =  int(x) + (x - int(x)) * 100 / 60
    #x = np.radians(x)
    return x
def float_to_deg(x:float):
    #x = np.degrees(x)
    x =  int(x) + (x - int(x)) * 60 / 100
    return x

def df_deg_to_float(df: pd.DataFrame) -> pd.DataFrame:
    df = df.apply(lambda x: x.astype(int) + (x - x.astype(int)) * 100 / 60)
    #df = df.apply(np.radians)
    return df
def df_float_to_deg(df: pd.DataFrame) -> pd.DataFrame:
    #df = df.apply(np.degrees)
    df = df.apply(lambda x: x.astype(int) + (x - x.astype(int)) * 60 / 100)
    return df
```

```python
# 原始数据
df = pd.DataFrame({
    'AB1':[59.32, 59.31, 59.33, 59.29, 59.32, 59.34],
    'AB2':[239.58, 239.48, 239.27, 239.28, 239.30, 239.31],
    'AC1':[298.31, 299.31, 299.31, 299.30, 299.32, 299.31],
    'AC2':[119.33, 119.34, 119.34, 119.29, 119.32, 119.33],
    'phi1':[0, 0, 0, 0, 0, 0],
    'phi2':[0, 0, 0, 0, 0, 0],
    'phi':[0, 0, 0, 0, 0, 0]
    })

df = df_deg_to_float(df)

df['phi1'] = df['AB1'] - df['AC1'] + 360 #np.radians(360)
df['phi2'] = df['AB2'] - df['AC2']
df['phi'] = 1/2 * (df['phi1'] + df['phi2'])

df = df_float_to_deg(df)

df['phi'] = round(df['phi'],2)
df
```

```python
# 计算平均值

f_df = df_deg_to_float(df)
m = f_df['phi'].mean()
m = float_to_deg(m)
print(m)

m = deg_to_float(m)
A = 180 - m
A = float_to_deg(A)
print(A)
```

```python
# 不确定度

s_phi = f_df['phi'].std()
print(s_phi)

d_phi = np.sqrt(s_phi**2 + 1)
print(d_phi)
d_phi = float_to_deg(d_phi)
print(d_phi)
```
