```python
import pandas as pd
import numpy as np

def deg_to_float(x:float):
    x =  int(x) + (x - int(x)) * 100 / 60
    #x = np.radians(x)
    return x
def float_to_deg(x:float):
    #x = np.degrees(x)
    x =  int(x) + (x - int(x)) * 60 / 100
    return x

def df_deg_to_float(df: pd.DataFrame) -> pd.DataFrame:
    df = df.apply(lambda x: x.astype(int) + (x - x.astype(int)) * 100 / 60)
    #df = df.apply(np.radians)
    return df
def df_float_to_deg(df: pd.DataFrame) -> pd.DataFrame:
    #df = df.apply(np.degrees)
    df = df.apply(lambda x: x.astype(int) + (x - x.astype(int)) * 60 / 100)
    return df
```

```python
df = pd.DataFrame({
    'lev':[50, 49, 48, 47, 46, 45, 20, 19, 18, 17, 16, 15],
    'left':[30.659, 30.600, 30.549, 30.496, 30.435, 30.372, 28.691, 28.610, 28.522, 28.438, 28.345, 28.252],
    'right':[19.375, 19.430, 19.482, 19.545, 19.590, 19.658, 21.339, 21.417, 21.502, 21.580, 21.681, 21.767],
    'd':[0] * 12,
    'dm2-dn2':[0] * 12
})

df['d'] = df['left'] - df['right']

dm2 = [df[df['lev'] == lev]['d'].to_numpy()**2 for lev in df['lev'] if lev >= 30]
dn2 = [df[df['lev'] == lev - 30]['d'].to_numpy()**2 for lev in df['lev'] if lev >= 30]
d_list = [float((dm2[i] - dn2[i]).squeeze()) for i in range(len(dm2))]
df['dm2-dn2'] = d_list + [0] * 6

df = df.apply(lambda x: round(x, 3))
df
```
