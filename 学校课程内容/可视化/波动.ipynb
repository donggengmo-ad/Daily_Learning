# 波动：行波与驻波

> 在三维空间 $(t, x, y)$ 中直观感受波动的传播与叠加

## 一、行波 (Traveling Wave)

### 物理图像

行波是**扰动在介质中传播**的波动形式。波形的每一个点都以相同的速度 $v$ 沿 $x$ 方向移动，能量随之传递。

### 数学表达式

沿 $+x$ 方向传播的简谐行波：

$$y(x, t) = A \sin(kx - \omega t + \varphi_0)$$

| 符号 | 含义 |
|------|------|
| $A$ | 振幅 |
| $k = 2\pi/\lambda$ | 波数（空间频率） |
| $\omega = 2\pi f = 2\pi/T$ | 角频率（时间频率） |
| $v = \omega/k$ | 波速（相速度） |
| $\varphi_0$ | 初相位 |

**关键特征**：波峰/波谷随时间沿 $x$ 方向平移，波形整体向前推进。

### 💡 降维：y(x,t) → f(φ)

行波 $y(x,t)=A\sin(kx-\omega t)$ 中 $x$ 和 $t$ 不是独立出现的——
它们通过**相位** $\varphi=kx-\omega t$ 组合在一起。令 $f(\varphi)=A\sin\varphi$，则：

$$\boxed{y(x,t)=f\big(\varphi(x,t)\big)}$$

#### 等相位线 ⊥ ψ 平面

定义两个正交方向（在 $x$-$t$ 平面内）：
- **相位方向** $\varphi = kx - \omega t$（波传播方向）
- **垂直方向** $\psi = \omega x + kt$（等相位线的走向，$(\omega,k)\cdot(k,-\omega)=0$）

固定 $\varphi$ → 等相位线（曲面上 $y$ 恒定的直线）。固定 $\psi$ → 垂直于所有等相位线的平面。

曲面与 $\psi=$const 平面的**交线**恰好是 $y(\varphi)=A\sin\varphi$——
三维曲面在这个平面上「收缩」为一条一维简谐曲线。

> 🎯 **几何本质**：行波曲面是**直纹面**——一族平行直线（等相位线）覆盖整个曲面。
> 沿等相位线方向看，不同 $(x,t)$ 只要 $\varphi$ 相同，$y$ 就相同。曲面退化。

下图：12 条彩色等相位线 + 淡蓝 ψ 平面扫描 + 橙红交线 $y(\varphi)$ 🎬

```python
# ========== 降维：等相位线与 ψ 平面扫掠 ==========
import numpy as np
import plotly.graph_objects as go

A = 1.0; lam = 2.0; T_period = 1.0
k = 2*np.pi/lam; omega = 2*np.pi/T_period; v = omega/k

x = np.linspace(0, 4*lam, 120)
t = np.linspace(0, 2*T_period, 80)
X, T = np.meshgrid(x, t)
Y = A * np.sin(k*X - omega*T)

k2w2 = k**2 + omega**2

# ---- 等相位线：覆盖整个可见 φ 区间 [-3π, 7π] ----
n_phi = 16
phi_vals = np.linspace(-3*np.pi, 7*np.pi, n_phi)
phi_colors = [
    '#E74C3C','#E67E22','#F39C12','#F1C40F','#2ECC71','#1ABC9C',
    '#3498DB','#2980B9','#9B59B6','#8E44AD','#E74C3C','#E67E22',
    '#F1C40F','#2ECC71','#3498DB','#9B59B6',
]

# ---- 初始 ψ 值 ----
psi_init = (psi_min := omega*x[0] + k*t[0])

# ---- 构建图 ----
fig_phi = go.Figure()

# ① 曲面
fig_phi.add_trace(go.Surface(
    x=X, y=T, z=Y, colorscale='RdBu', cmin=-A, cmax=A,
    opacity=0.4, showscale=False, name='y(x,t)',
))

# ②-⑰ 等相位线
for phi, clr in zip(phi_vals, phi_colors):
    t_line = np.linspace(0, 2*T_period, 200)
    x_line = (phi + omega*t_line) / k
    z_line = np.full_like(t_line, A*np.sin(phi))
    mask = (x_line >= 0) & (x_line <= 4*lam)
    fig_phi.add_trace(go.Scatter3d(
        x=x_line[mask], y=t_line[mask], z=z_line[mask],
        mode='lines', line=dict(width=2.5, color=clr),
        name=f'φ={phi/np.pi:.1f}π', showlegend=bool(abs(phi) < 1e-10),
    ))

# ⑱ ψ 平面（初始）
t_corners = np.array([0, 2*T_period, 2*T_period, 0])
x_corners = np.clip((psi_init - k*t_corners) / omega, 0, 4*lam)
y_corners = np.array([-1.5*A, -1.5*A, 1.5*A, 1.5*A])
fig_phi.add_trace(go.Mesh3d(
    x=x_corners, y=t_corners, z=y_corners,
    color='lightblue', opacity=0.2, name='ψ 平面',
))

# ⑲ 交线 y(φ)（初始）
phi_param = np.linspace(-4*np.pi, 8*np.pi, 400)
x_param = (k*phi_param + omega*psi_init) / k2w2
t_param = (-omega*phi_param + k*psi_init) / k2w2
y_param = A * np.sin(phi_param)
mask_param = (x_param >= 0) & (x_param <= 4*lam) & (t_param >= 0) & (t_param <= 2*T_period)
fig_phi.add_trace(go.Scatter3d(
    x=x_param[mask_param], y=t_param[mask_param], z=y_param[mask_param],
    mode='lines', line=dict(width=6, color='#FF4500'),
    name='y(φ) 降维',
))

# ====== 帧 ======
n_frames = 25
psi_max = omega*x[-1] + k*t[-1]
psi_frames = np.linspace(psi_min, psi_max, n_frames)

frames_phi = []
for psi_val in psi_frames:
    data = [go.Surface(
        x=X, y=T, z=Y, colorscale='RdBu', cmin=-A, cmax=A,
        opacity=0.4, showscale=False,
    )]

    # 等相位线
    for phi, clr in zip(phi_vals, phi_colors):
        t_line = np.linspace(0, 2*T_period, 200)
        x_line = (phi + omega*t_line) / k
        z_line = np.full_like(t_line, A*np.sin(phi))
        mask = (x_line >= 0) & (x_line <= 4*lam)
        data.append(go.Scatter3d(
            x=x_line[mask], y=t_line[mask], z=z_line[mask],
            mode='lines', line=dict(width=2.5, color=clr), showlegend=False,
        ))

    # ψ 平面
    x_c = np.clip((psi_val - k*t_corners) / omega, 0, 4*lam)
    data.append(go.Mesh3d(
        x=x_c, y=t_corners, z=y_corners,
        color='lightblue', opacity=0.2, showlegend=False,
    ))

    # 交线
    x_p = (k*phi_param + omega*psi_val) / k2w2
    t_p = (-omega*phi_param + k*psi_val) / k2w2
    y_p = A * np.sin(phi_param)
    m_p = (x_p >= 0) & (x_p <= 4*lam) & (t_p >= 0) & (t_p <= 2*T_period)
    data.append(go.Scatter3d(
        x=x_p[m_p], y=t_p[m_p], z=y_p[m_p],
        mode='lines', line=dict(width=6, color='#FF4500'), showlegend=False,
    ))

    frames_phi.append(go.Frame(data=data, name=f'psi={psi_val:.1f}'))

fig_phi.update(frames=frames_phi)

fig_phi.update_layout(
    title=dict(
        text=f'<b>降维</b>  ψ = ωx+kt 平面扫过曲面  →  交线 = y(φ)=A·sin(φ)<br>'
             f'<sup>{n_phi} 条等相位线（彩色，覆盖 φ∈[-3π,7π]）  |  淡蓝 ψ 平面  |  橙红交线</sup>',
        x=0.5, xanchor='center',
    ),
    width=900, height=650,
    margin=dict(l=0, r=0, t=70, b=0),
    legend=dict(x=0.02, y=0.98),
    transition=dict(duration=0, easing='linear'),
    updatemenus=[dict(
        type='buttons', direction='left',
        x=0.1, xanchor='right', y=0, yanchor='top',
        buttons=[
            dict(label='▶', method='animate',
                 args=[None, dict(frame=dict(duration=100, redraw=True),
                                  fromcurrent=False, mode='immediate',
                                  transition=dict(duration=0))]),
            dict(label='⏸', method='animate',
                 args=[[None], dict(frame=dict(duration=0, redraw=False),
                                    mode='immediate')]),
        ],
    )],
    sliders=[dict(active=0, len=0.9, x=0.05, pad=dict(t=20),
                  currentvalue=dict(prefix='ψ = ', font=dict(size=14)),
                  steps=[dict(method='animate',
                              args=[[f'psi={pv:.1f}'], dict(mode='immediate',
                                    frame=dict(duration=100, redraw=True))],
                              label=f'{pv:.1f}') for pv in psi_frames])],
)
fig_phi.update_scenes(
    xaxis_title='x (m)', yaxis_title='t (s)', zaxis_title='y (m)',
    camera=dict(eye=dict(x=1.4, y=-1.8, z=1.0)), aspectratio=dict(x=2, y=1, z=0.8),
)

fig_phi.show()
```

## 二、驻波 (Standing Wave)

### 物理图像

驻波是两列**同频率、同振幅、反向传播**的行波叠加的结果。

$$\begin{aligned}
y_1 &= A \sin(kx - \omega t) \quad &\text{（向右）} \\
y_2 &= A \sin(kx + \omega t) \quad &\text{（向左）} \\
y &= y_1 + y_2 = 2A \sin(kx) \cos(\omega t)
\end{aligned}$$

### 驻波方程

$$\boxed{y(x, t) = 2A \sin(kx) \cos(\omega t)}$$

| 特征 | 说明 |
|------|------|
| **波节 (Node)** | $\sin(kx) = 0$ 的位置，始终静止，$x = n\lambda/2$ |
| **波腹 (Antinode)** | $|\sin(kx)| = 1$ 的位置，振幅最大，$x = (2n+1)\lambda/4$ |
| **能量** | 不沿 x 方向传播，只在波节与波腹之间来回转换 |
| **相位** | 相邻波节之间各点同相振动；波节两侧反相 |

> 💡 **关键洞察**：驻波方程是一个**分离变量**形式——空间部分 $\sin(kx)$ 和时间部分 $\cos(\omega t)$ 相乘。这意味着空间波形和时间演化被「解耦」了，不同位置的振动模式固定在空间框架上。

### 💻 驻波 — 3D 联动视图


```python
# ========== 驻波：3D 曲面 + 2D 波形联动 ==========
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

A = 1.0; lam = 2.0; T_period = 1.0
k = 2*np.pi/lam; omega = 2*np.pi/T_period

x = np.linspace(0, 4*lam, 120)
t = np.linspace(0, 2*T_period, 80)
X, T = np.meshgrid(x, t)
Y = 2*A * np.sin(k*X) * np.cos(omega*T)

x_dense = np.linspace(0, 4*lam, 300)
n_frames = 40
t_frames = np.linspace(0, T_period, n_frames)

n_lo = 50, 30
x_lo = np.linspace(0, 4*lam, n_lo[0])
t_lo = np.linspace(0, 2*T_period, n_lo[1])
X_lo, T_lo = np.meshgrid(x_lo, t_lo)
Y_lo = 2*A * np.sin(k*X_lo) * np.cos(omega*T_lo)

node_x = np.arange(0, 4*lam + 0.01, lam/2)
n_nodes = len(node_x)

fig_sw = make_subplots(
    rows=2, cols=1,
    subplot_titles=(
        '<b>3D</b>  y(x,t) = 2A·sin(kx)·cos(wt)',
        '<b>2D</b>  波形快照 y(x)'
    ),
    specs=[[{'type': 'scene'}], [{'type': 'xy'}]],
    row_heights=[0.68, 0.32], vertical_spacing=0.01,
)

fig_sw.add_trace(go.Surface(
    x=X, y=T, z=Y, colorscale='RdBu',
    cmin=-2*A, cmax=2*A,
    opacity=1.0, showscale=False, name='曲面',
), row=1, col=1)

for nx in node_x:
    fig_sw.add_trace(go.Scatter3d(
        x=[nx, nx], y=[0, 2*T_period], z=[0, 0],
        mode='lines', line=dict(width=2, color='gray', dash='dash'),
        showlegend=False,
    ), row=1, col=1)

fig_sw.add_trace(go.Scatter3d(
    x=x_dense, y=np.full_like(x_dense, t_frames[0]),
    z=2*A*np.sin(k*x_dense)*np.cos(omega*t_frames[0]),
    mode='lines', line=dict(width=5, color='gold'),
    name='截面 y(x)', visible=False,
), row=1, col=1)

fig_sw.add_trace(go.Scatter(
    x=x, y=2*A*np.sin(k*x)*np.cos(omega*t_frames[0]),
    mode='lines', line=dict(width=3, color='#9D3DCF'),
    name='y(x)', visible=False,
), row=2, col=1)

frames = []
for tv in t_frames:
    data = [go.Surface(x=X_lo, y=T_lo, z=Y_lo,
                       colorscale='RdBu', cmin=-2*A, cmax=2*A,
                       opacity=0.4, showscale=False)]
    for nx in node_x:
        data.append(go.Scatter3d(
            x=[nx, nx], y=[0, 2*T_period], z=[0, 0],
            mode='lines', line=dict(width=2, color='gray', dash='dash'),
            showlegend=False))
    data.append(go.Scatter3d(
        x=x_dense, y=np.full_like(x_dense, tv),
        z=2*A*np.sin(k*x_dense)*np.cos(omega*tv),
        mode='lines', line=dict(width=5, color='gold')))
    data.append(go.Scatter(
        x=x, y=2*A*np.sin(k*x)*np.cos(omega*tv),
        mode='lines', line=dict(width=3, color='#9D3DCF')))
    frames.append(go.Frame(data=data, name=f't={tv:.2f}'))

tv0 = t_frames[0]
data_loop = [go.Surface(x=X_lo, y=T_lo, z=Y_lo,
                         colorscale='RdBu', cmin=-2*A, cmax=2*A,
                         opacity=0.4, showscale=False)]
for nx in node_x:
    data_loop.append(go.Scatter3d(
        x=[nx, nx], y=[0, 2*T_period], z=[0, 0],
        mode='lines', line=dict(width=2, color='gray', dash='dash'),
        showlegend=False))
data_loop.append(go.Scatter3d(
    x=x_dense, y=np.full_like(x_dense, tv0),
    z=2*A*np.sin(k*x_dense)*np.cos(omega*tv0),
    mode='lines', line=dict(width=5, color='gold')))
data_loop.append(go.Scatter(
    x=x, y=2*A*np.sin(k*x)*np.cos(omega*tv0),
    mode='lines', line=dict(width=3, color='#9D3DCF')))
frames.append(go.Frame(data=data_loop, name='t=loop'))

fig_sw.update(frames=frames)

fig_sw.update_layout(
    title_text=f'<b>驻波</b>  lam={lam}m  T={T_period}s  波节间距 lam/2={lam/2}m',
    width=900, height=650,
    margin=dict(l=0, r=0, t=50, b=0),
    transition=dict(duration=0, easing='linear'),
    updatemenus=[
        dict(type='buttons', direction='right',
             x=0.0, y=1.08, xanchor='left', yanchor='bottom',
             buttons=[
                 dict(label='纯 3D', method='update',
                      args=[{'visible':  [True]* (1+n_nodes) + [False, False],
                             'opacity':  [1.0]* (1+n_nodes) + [1.0,  1.0]},
                            {'scene.domain.y':      [0.02, 0.98],
                             'yaxis.domain':        [0, 0],
                             'xaxis.visible':       False,
                             'yaxis.visible':       False}]),
                 dict(label='联动', method='update',
                      args=[{'visible':  [True]* (1+n_nodes) + [True,  True],
                             'opacity':  [0.4]*1+[1.0]*n_nodes + [1.0,  1.0]},
                            {'scene.domain.y':      [0.35, 0.98],
                             'yaxis.domain':        [0.05, 0.28],
                             'xaxis.visible':       True,
                             'yaxis.visible':       True}]),
             ]),
        dict(type='buttons', direction='left',
             x=0.1, xanchor='right', y=0, yanchor='top',
             buttons=[
                 dict(label='▶', method='animate',
                      args=[None, dict(frame=dict(duration=60, redraw=True),
                                       fromcurrent=False,
                                       mode='immediate',
                                       transition=dict(duration=0))]),
                 dict(label='⏸', method='animate',
                      args=[[None], dict(frame=dict(duration=0, redraw=False),
                                         mode='immediate')]),
             ]),
    ],
    sliders=[dict(active=0, len=0.9, x=0.05, pad=dict(t=20),
                  currentvalue=dict(prefix='t = ', suffix=' s', font=dict(size=14)),
                  steps=[dict(method='animate',
                              args=[[f't={tv:.2f}'], dict(mode='immediate',
                                    frame=dict(duration=60, redraw=True))],
                              label=f'{tv:.2f}') for tv in t_frames])],
)
fig_sw.update_scenes(
    xaxis_title='x (m)', yaxis_title='t (s)', zaxis_title='y (m)',
    camera=dict(eye=dict(x=1.6, y=-1.8, z=1.2)), aspectratio=dict(x=2, y=1, z=0.8),
)
fig_sw.update_xaxes(title_text='x (m)', range=[0, 4*lam], dtick=lam/2, row=2, col=1)
fig_sw.update_yaxes(title_text='y (m)', range=[-2.5*A, 2.5*A], row=2, col=1)

fig_sw.show()

```

### 读图要点

- **波节线**：沿 x 轴等间距分布，可以看到曲面在这些位置始终贴近 $y=0$ 平面——这是「不动的节点」。
- **波腹振荡**：波节之间的区域在正负最大位移之间来回振荡，这是振幅最大的地方。
- **与行波对比**：注意到曲面没有沿 x 方向的「倾斜脊线」——波形不传播，只在原地「呼吸」。
- **时变模式**：$t = T/4, 3T/4$ 时 $\cos(\omega t) = 0$，整条弦瞬间通过平衡位置（曲面变平）。

## 三、总结

| 维度 | 行波 | 驻波 |
|------|------|------|
| **方程** | $A\sin(kx \mp \omega t)$ | $2A\sin(kx)\cos(\omega t)$ |
| **变量耦合** | $x$ 和 $t$ 耦合（组合变量） | $x$ 和 $t$ 分离 |
| **能量** | 沿传播方向传输 | 不做定向传输 |
| **波形** | 整体平移 | 原地振荡 |
| **振幅** | 处处相同 ($A$) | 随空间变化 ($0 \sim 2A$) |
| **3D 曲面特征** | 倾斜脊线，沿 +x 走 | 竖直脊线，原地「呼吸」 |

> 🎯 **一句话记住**：行波是「波形在跑」，驻波是「振幅在抖」——前者 $x,t$ 绑在一起走，后者 $x,t$ 各玩各的。

> 🔗 驻波 = 两列反向行波叠加 → 干涉产生稳定的空间能量分布，是理解**共振、乐器、量子定态**的基础。
