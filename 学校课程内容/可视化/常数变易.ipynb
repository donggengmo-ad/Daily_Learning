# 常数变易法 (Variation of Constants)

> 从齐次解的曲线族出发，把常数「变活」——一阶线性非齐次方程的通用解法

---


## 一、问题

一阶线性非齐次方程：

$$\frac{dy}{dx} + P(x)\,y = Q(x)$$

### Step 1 — 齐次通解

先解对应的齐次方程 $y' + P(x)y = 0$：

$$y_h = C \cdot e^{-\int P(x)\,dx}$$

这是一族曲线，$C$ 取不同值得到不同的积分曲线。

### Step 2 — 常数「变易」

**核心想法**：把常数 $C$ 换成一个未知函数 $C(x)$，令 $y = C(x) \cdot e^{-\int P dx}$，代入原非齐次方程：

$$C'(x) \cdot e^{-\int P dx} = Q(x) \quad\Rightarrow\quad C'(x) = Q(x) \cdot e^{\int P dx}$$

积分得 $C(x) = \int Q(x) e^{\int P dx}\,dx + K$。

### Step 3 — 通解

$$\boxed{y = \underbrace{K \cdot e^{-\int P dx}}_{\text{齐次通解}} + \underbrace{e^{-\int P dx}\int Q e^{\int P dx}\,dx}_{\text{一个特解}}}$$

下面以 $y' + y = x$ 为例可视化整个过程。


```python
# ========== 齐次方程 y' + y = 0 -> y = C·e^{-x} ==========
import numpy as np
import plotly.graph_objects as go

x = np.linspace(-2, 6, 400)
C_vals = np.arange(-2, 4)
colors = ['#3498DB','#2980B9','#1ABC9C','#95A5A6','#E67E22','#E74C3C']
C_highlight = 2

fig_h = go.Figure()

for C, clr in zip(C_vals, colors):
    y = C * np.exp(-x)
    is_hl = bool(C == C_highlight)
    fig_h.add_trace(go.Scatter(
        x=x, y=y, mode='lines',
        line=dict(width=3.5 if is_hl else 1.5,
                  color='#F1C40F' if is_hl else clr),
        showlegend=False,
    ))

# 坐标轴
fig_h.add_hline(y=0, line=dict(width=1.5, color='#555', dash='solid'))
fig_h.add_vline(x=0, line=dict(width=1.5, color='#555', dash='solid'))

# 标注
fig_h.add_annotation(x=1.5, y=2*np.exp(-1.5)+0.3,
    text='C=2', showarrow=False,
    font=dict(size=14, color='#F1C40F'))

fig_h.update_layout(
    title='<b>齐次方程</b>  y\' + y = 0  ->  y = C·e<sup>-x</sup>',
    xaxis=dict(title='x', range=[-2, 6]),
    yaxis=dict(title='y', range=[-3, 5]),
    width=700, height=450,
    margin=dict(l=50, r=20, t=50, b=50),
    showlegend=False,
)

fig_h.show()
```

## 二、常数变易

对 $y' + y = x$，齐次通解 $y_h = C e^{-x}$。令 $C \to C(x)$ 代入非齐次方程：

$$C'(x)e^{-x} = x \quad\Rightarrow\quad C'(x) = x e^x$$

积分：$C(x) = \int x e^x\,dx = (x-1)e^x + K$。

非齐次通解：

$$\boxed{y = K e^{-x} + (x-1)}$$

齐次族中 $C=2$ 那条曲线，经过常数变易后变成 $K=2$ 的曲线（常量 $K$ 取相同值）。
下面是变易前后的对比——注意观察 $C=2$ 那条线如何被 $(x-1)$ 这一项「顶起来」。


```python
# ========== 非齐次 y' + y = x -> y = K·e^{-x} + (x-1) ==========
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

x = np.linspace(-2, 6, 400)
C_vals = np.arange(-2, 4)
colors = ['#3498DB','#2980B9','#1ABC9C','#95A5A6','#E67E22','#E74C3C']
K_highlight = 2

fig = make_subplots(rows=1, cols=2,
    subplot_titles=(
        '<b>变易前</b>  y<sub>h</sub> = C·e<sup>-x</sup>',
        '<b>变易后</b>  y = K·e<sup>-x</sup> + (x-1)'
    ),
    horizontal_spacing=0.08,
)

# ---- 左：齐次族 ----
for C, clr in zip(C_vals, colors):
    y = C * np.exp(-x)
    hl = bool(C == K_highlight)
    fig.add_trace(go.Scatter(
        x=x, y=y, mode='lines',
        line=dict(width=3.5 if hl else 1.5,
                  color='#F1C40F' if hl else clr),
        showlegend=False,
    ), row=1, col=1)

# ---- 右：非齐次族 ----
for K, clr in zip(C_vals, colors):
    y = K * np.exp(-x) + (x - 1)
    hl = bool(K == K_highlight)
    fig.add_trace(go.Scatter(
        x=x, y=y, mode='lines',
        line=dict(width=3.5 if hl else 1.5,
                  color='#F1C40F' if hl else clr),
        showlegend=False,
    ), row=1, col=2)

# 右图：渐近线 y = x-1（和坐标轴同色）
fig.add_trace(go.Scatter(
    x=x, y=x-1, mode='lines',
    line=dict(width=2, color='#555', dash='dashdot'),
    showlegend=False,
), row=1, col=2)

# 坐标轴
for col in [1, 2]:
    fig.add_hline(y=0, line=dict(width=1.5, color='#555', dash='solid'), row=1, col=col)
    fig.add_vline(x=0, line=dict(width=1.5, color='#555', dash='solid'), row=1, col=col)

# 右图标注
fig.add_annotation(x=4.5, y=2*np.exp(-4.5)+4.5-1+0.3,
    text='K=2 变易后', showarrow=False,
    font=dict(size=13, color='#F1C40F'), row=1, col=2)

fig.update_layout(
    title_text='<b>常数变易</b>  齐次解 C·e<sup>-x</sup>  ->  非齐次解 K·e<sup>-x</sup> + (x-1)',
    width=1100, height=450,
    margin=dict(l=50, r=20, t=60, b=50),
    showlegend=False,
)
fig.update_xaxes(title_text='x', range=[-2, 6])
fig.update_yaxes(title_text='y', range=[-3, 6], row=1, col=1)
fig.update_yaxes(title_text='y', range=[-3, 6], row=1, col=2)

fig.show()
```

---

## 三、总结

| 步骤 | 操作 | 本例 |
|------|------|------|
| ① 齐次通解 | $y_h = C \cdot e^{-\int P dx}$ | $y_h = C e^{-x}$ |
| ② 常数变易 | 令 $C \to C(x)$，代入非齐次求 $C'(x)$ | $C'(x) = x e^x$ |
| ③ 积分得 $C(x)$ | $C(x) = \int Q e^{\int P dx}\,dx + K$ | $C(x) = (x-1)e^x + K$ |
| ④ 通解 | $y = C(x) e^{-\int P dx}$ | $y = K e^{-x} + (x-1)$ |

> 🎯 **几何直觉**：齐次方程的解是一族「形状相同、按比例缩放」的曲线。
> 非齐次项 $Q(x)$ 给每条曲线加上了一个**共同的垂直偏移**——
> 曲线族的「形状」不变，但整个族在 $y$ 方向被扭了一下。
