```python
import torch
import numpy as np
from torch.autograd import Variable
```

```python
# numpy与tensor的对转
np_data = np.arange(6).reshape((2, 3))
print(np_data)

torch_data = torch.from_numpy(np_data)
print(torch_data)

re_np_data = torch_data.numpy()
print(re_np_data)
```

```python
# tensor操作方法
data = [[-1, -2], [1, 2]]
tensor = torch.FloatTensor(data)

print(tensor.abs())
print(tensor.sin())
print(torch.mm(tensor, tensor)) # matrix multiplication
```

```python
# variable
tensor = torch.FloatTensor([[1, 2], [3, 4]])
v = Variable(tensor, requires_grad=True)
print(v.data)

t_out = torch.mean(v ** 3)
t_out.backward()

print(v.grad)
```
