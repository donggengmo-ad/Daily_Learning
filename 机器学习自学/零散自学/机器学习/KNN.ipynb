```python
import mglearn
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
import numpy as np
import matplotlib.pyplot as plt
```

## KNN算法
- 根据距离度量找到最近的K个邻居
- 根据这些邻居的类别进行投票来确定新数据点的类别

```python
plt.subplots(1,2,figsize=(10,5))
plt.subplot(1,2,1)
mglearn.plots.plot_knn_classification(n_neighbors=1)
plt.subplot(1,2,2)
mglearn.plots.plot_knn_classification(n_neighbors=5)
```

## K值选择
K值的选择会影响决策边界的复杂度和模型的性能
- 小K值 决策边界复杂，容易过拟合
- 大K值 决策边界简单，容易欠拟合

```python
X, y = mglearn.datasets.make_forge()
fig, axes = plt.subplots(1,3,figsize=(12,4))
for n, ax in zip([1,3,9], axes):
    knn = KNeighborsClassifier(n_neighbors=n).fit(X, y)
    mglearn.plots.plot_2d_classification(knn,X_train,y_train,alpha=0.4,ax=ax)
    mglearn.discrete_scatter(X[:,0], X[:,1], y ,ax=ax)
    ax.set_title(f"{n} neighbors")
axes[0].legend(loc=3)
```

## KNN回归
根据最近的K个邻居的平均值来预测新数据点的数值

```python
line = np.linspace(-3, 3, 1000).reshape(-1,1)
X = np.linspace(-3, 3, 20).reshape(-1,1)
y = np.sin(X)
fig, axes = plt.subplots(1,3,figsize=(12,4))
for n, ax in zip([1,3,9], axes):
    knn = KNeighborsRegressor(n_neighbors=n).fit(X, y)
    ax.plot(line, knn.predict(line), color='red', alpha=0.5)
    ax.plot(X, y, '^', color='blue', alpha=0.5)
    ax.set_title(f"{n} neighbors")
    ax.set_title(f"{n} neighbors")
```
