```python
import torch
from torch import nn
import util
```

```python
batch_size = 32
num_steps = 35
train_iter, vocab = util.data.load_data_songci(batch_size, num_steps)
```

```python
vocab_size = len(vocab)
num_hiddens = 256
num_layers = 4
device = util.try_gpu()
num_epochs = 100
lr = 2
```

```python
gru_layer = nn.GRU(input_size=vocab_size, hidden_size=num_hiddens)
model = util.model.RNNModel(gru_layer, vocab_size).to(device)
util.train.train_ch8(model, train_iter, vocab, lr, num_epochs, device, prefix=['春花秋月何时了，往事知多少。'])
```

```python
util.predict_ch8('春花秋月何时了，往事知多少。', 175, model, vocab, device)
```
